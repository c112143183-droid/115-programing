#九九乘法表

1\*1=1, 1\*2=2, 1\*3=3,...
2\*1=2, 2\*2=4, 2\*3=6,...
3\*1=3, 3\*2=6, 3\*3=9,...

```python
print(f'1*1={1*1}')
print(f'1*2={1*2}')
print(f'1*3={1*3}')
```

```python
print(f'1*1={1*1}',end=',')
print(f'1*2={1*2}',end=',')
```

```python
j=1
print(f'1*{j}={1*j}',end=',')
j=2
print(f'1*{j}={1*j}',end=',')
j=3
print(f'1*{j}={1*j}',end=',')
```

```python
for j in range(1,6):
    print(f'1*{j}={1*j}',end=',')
print()
for j in range(1,6):
    print(f'2*{j}={2*j}',end=',')
print()
for j in range(1,6):
    print(f'3*{j}={3*j}',end=',')
print()
for j in range(1,6):
    print(f'4*{j}={4*j}',end=',')
print()
for j in range(1,6):
    print(f'5*{j}={5*j}',end=',')
print()
for j in range(1,6):
    print(f'6*{j}={6*j}',end=',')
print()
for j in range(1,6):
    print(f'7*{j}={7*j}',end=',')
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```

# 練習:印出質數

質數:一個數字n，除了自己本身，以及1以外，沒有任何樹可以除盡他，n就是質數。

```python
n=273
i=2

if 273 % 2 == 0:
    print(f"{273}不是質數")
if 273 % 3 == 0:
    print(f"{273}不是質數")
if 273 % 4 == 0:
    print(f"{273}不是質數")
```

```python
for i in range(2,354257347347147575713):
    if 354257347347147575713 % i==0:
        print(f"{354257347347147575713}不是質數，因為它可以被{i}除盡")
        break #只要數字能被除盡，就脫離迴圈
```

```python
n=44657627353715
isPrime =True
for i in range(2,n):
    if n % i==0:
        print(f"{n}不是質數，因為它可以被{i}除盡")
        isPrime =False
        break #只要數字能被除盡，就脫離迴圈

if isPrime == True:
    print(f'{n}是質數')
```
