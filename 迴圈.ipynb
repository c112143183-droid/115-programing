#九九乘法表

1\*1=1, 1\*2=2, 1\*3=3,...
2\*1=2, 2\*2=4, 2\*3=6,...
3\*1=3, 3\*2=6, 3\*3=9,...

```python
print(f'1*1={1*1}')
print(f'1*2={1*2}')
print(f'1*3={1*3}')
```

```python
print(f'1*1={1*1}',end=',')
print(f'1*2={1*2}',end=',')
```

```python
j=1
print(f'1*{j}={1*j}',end=',')
j=2
print(f'1*{j}={1*j}',end=',')
j=3
print(f'1*{j}={1*j}',end=',')
```

```python
for j in range(1,6):
    print(f'1*{j}={1*j}',end=',')
print()
for j in range(1,6):
    print(f'2*{j}={2*j}',end=',')
print()
for j in range(1,6):
    print(f'3*{j}={3*j}',end=',')
print()
for j in range(1,6):
    print(f'4*{j}={4*j}',end=',')
print()
for j in range(1,6):
    print(f'5*{j}={5*j}',end=',')
print()
for j in range(1,6):
    print(f'6*{j}={6*j}',end=',')
print()
for j in range(1,6):
    print(f'7*{j}={7*j}',end=',')
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```
