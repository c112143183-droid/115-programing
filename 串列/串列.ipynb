# 串列是一個可以儲存多個元素(整數，浮點數，或字串)的容器
# 串列中每個元素都是使用逗點隔開，左右有中括號[]。

```python
# scores 是一個串列，裏面每個元素都是整數。
scores=[78,29,56,80,99]
print(scores)
```

# 我們可以使用索引值(index)來存取列中的元素。
# 注意：索引值由0開始。

```python
print(scores[0]) #印出第1個元素
print(scores[1]) #印出第2個元素
```

# 串列內容的個數
len(a)取得a這個串列中元素的個數。
len是length(長度)的意思

```python
print(len(scores))
print(scores[4]) #取得最後一個元素
print(scores[4]) #取得最後一個元素
print(scores[len(scores)-1])
print(scores[-1])

```

```python
print(scores)
scores.append(78)
scores.append(29)
scores.append(56)
scores.append(89)
scores.append(99)
scores.append(73)
```

## 走訪串列的三種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
#對於 `scores` 中的每個元素，一次取一個，放在a中
for a in scores:
    print(a)
```

```python
for i, a in enumerate(scores):
    print(i, a)
```

一個費伯納西數列如下:
1,1,2,3,5,8,...
請問第91個元素為何？
提示:
令$x$是一個費伯納西數列, $x[1]$是第 $x$個數字。
我們會得到 $x[n]=x[n-1]+x[n-2]$

```python
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
fib[2]=fib[0]+fib[1]
print(fib)
fib[3]=fib[1]+fib[2]
print(fib)
fib[4]=fib[2]+fib[3]
```

```python
for i in range(2,91):
    fib[i]=fib[i-2]+fib[i-1]

print(fib[90])
```
